## 1. My lane as an ML task (type)

Task Type: Ranking / Scoring (Learning-to-Rank / Regression)

Why: The objective is to evaluate how effectively a piece of content answers a user query compared to competitor pages on the Search Engine Results Page (SERP). Ranking or continuous scoring allows us to model relative relevance, topical depth, and quality signals across multiple competing documents rather than treating it as a simple binary outcome.

```python
# Verify class/task representation
task_type = "ranking_and_scoring"
print(f"Configured Task Type: {task_type}")
```

## 2. Target or proxy

Target / Proxy: Target SERP Rank Position ($1$ to $10$) or Relative Visibility Score ($0.0$ to $1.0$).Origin: This label comes from observed outcomes—measured search engine result rankings for a target query over a 30-day window after indexing.

```python
# Target variable definition
target_col = "serp_rank"
proxy_col = "relative_visibility_score"
print(f"Primary Target: {target_col} | Proxy Target: {proxy_col}")
```

## 3. Success metric

Offline Metric: NDCG@10 (Normalized Discounted Cumulative Gain at top 10) $\ge 0.75$ and Spearman’s Rank Correlation $\ge 0.65$.Why this number: An NDCG@10 above $0.75$ ensures the model accurately orders high-quality, highly relevant content near the top positions, reflecting true search ranking behavior.

```python
# Defining target offline benchmarks
success_metrics = {
    "NDCG@10_target": 0.75,
    "Spearman_rho_target": 0.65
}
for metric, benchmark in success_metrics.items():
    print(f"{metric}: >= {benchmark}")
```

## 4. The unit of analysis, as a real dataframe

Unit of Analysis: 1 Row = 1 (Target Query, Document URL) pair.

Each row represents a specific page evaluated against a given keyword or search query, containing lexical, semantic, and structural signals alongside the observed rank.

```python
import pandas as pd
import numpy as np

# Load starter data slice or define mock structure if running standalone
try:
    # Adjust path if data exists in your repo
    df = pd.read_parquet("../data/starter_data.parquet")
except Exception:
    # Representative schema for the lane
    df = pd.DataFrame({
        "query_id": ["q101", "q101", "q102", "q102"],
        "doc_id": ["d001", "d002", "d003", "d004"],
        "word_count": [1450, 2100, 850, 1920],
        "semantic_similarity": [0.88, 0.72, 0.65, 0.91],
        "serp_rank": [1, 4, 8, 2]
    })

print(f"DataFrame Shape: {df.shape}")
print("Unit of Analysis: 1 row = 1 (Query, Document) pair")
df.head()
```

## 5. Why ML beats a fixed rule here

A fixed heuristic (such as if word_count > 1500 and keyword_count > 5 then rank = top 3) fails because search intent varies widely across topics. Some queries favor concise, schema-rich answers, while others require comprehensive long-form coverage.

ML captures non-linear interactions across topical completeness, readability, semantic embedding similarity, and structural cues that cannot be captured by rigid threshold rules.

```python
# Check non-linear distribution between content size and rank
correlation = df[["word_count", "serp_rank"]].corr().iloc[0, 1]
print(f"Linear correlation between word count and rank: {correlation:.2f} (shows simple rules don't suffice)")
```
