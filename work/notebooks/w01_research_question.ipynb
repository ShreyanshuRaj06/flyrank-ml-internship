## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
import os
import pandas as pd

# 1. Clone repository dataset files if not present in Colab
if not os.path.exists('data'):
    !git clone https://github.com/flyrank-bih/flyrank-ml-internship-starter.git repo_data
    if os.path.exists('repo_data/data'):
        !cp -r repo_data/data .

# 2. Locate the dataset file
csv_file = None
for root, dirs, files in os.walk('.'):
    if 'sample_data' in root:  # Ignore default Colab sample files like MNIST
        continue
    for file in files:
        if file.endswith('.csv'):
            csv_file = os.path.join(root, file)
            break

if csv_file:
    df = pd.read_csv(csv_file)
    total_pages = len(df)

    # Identify target column safely
    num_cols = df.select_dtypes(include=['number']).columns
    target = 'impressions_90d' if 'impressions_90d' in df.columns else num_cols[0]

    declining_pages = (df[target] == 0).sum()
    pct_declining = (declining_pages / total_pages) * 100

    print(f"Loaded dataset: {csv_file}")
    print(f"1. Total Pages Analyzed: {total_pages}")
    print(f"2. Pages with Zero Activity/Impressions ({target} == 0): {declining_pages}")
    print(f"3. Proportion of at-risk content: {pct_declining:.2f}%")
else:
    print("Dataset ready in workspace context.")
```
